```python
cols = list(df.columns)

def find_cols(keywords):
    out = []
    for c in cols:
        low = c.lower()
        if any(k in low for k in keywords):
            out.append(c)
    return out

print("Possible image columns:", find_cols(["image", "img", "rgb", "camera"]))
print("Possible beam columns:", find_cols(["beam", "label", "index", "power"]))
print("Possible lat columns:", find_cols(["lat"]))
print("Possible lon columns:", find_cols(["lon", "lng", "long"]))
print("Possible episode columns:", find_cols(["episode", "scene", "unit", "run", "traj"]))
```

```python
CONFIG = {
    "root": str(root),
    "csv_path": str(CSV_PATH),

    "image_col": "unit1_rgb",
    "label_col": "unit1_beam_index",
    "episode_col": "seq_index",

    "bs_loc_col": "unit1_loc",
    "uav_loc_col": "unit2_loc",

    "image_size": 256,
    "seq_len": 4,
    "codebook_size": 64,

    # DeepSense beam index usually 1..64 হলে True লাগবে
    "label_is_one_indexed": True,
}

print(CONFIG)
```

```python
from pathlib import Path
from typing import List, Tuple, Dict
import pandas as pd
from PIL import Image

import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms


def resolve_path(root: Path, value: str) -> Path:
    p = Path(str(value))
    if p.is_absolute():
        return p
    return root / p


class UAVBeamSequenceDataset(Dataset):
    def __init__(
        self,
        df: pd.DataFrame,
        root: str,
        image_col: str,
        label_col: str,
        gps_cols: Tuple[str, str],
        seq_indices: List[List[int]],
        image_size: int = 256,
        label_is_one_indexed: bool = False,
    ):
        self.df = df.reset_index(drop=True)
        self.root = Path(root)
        self.image_col = image_col
        self.label_col = label_col
        self.gps_cols = gps_cols
        self.seq_indices = seq_indices
        self.label_is_one_indexed = label_is_one_indexed

        self.transform = transforms.Compose([
            transforms.Resize((image_size, image_size)),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225],
            ),
        ])

    def __len__(self):
        return len(self.seq_indices)

    def _load_image(self, row):
        path = resolve_path(self.root, row[self.image_col])
        img = Image.open(path).convert("RGB")
        return self.transform(img)

    def __getitem__(self, idx):
        ids = self.seq_indices[idx]
        rows = self.df.iloc[ids]

        images = torch.stack(
            [self._load_image(row) for _, row in rows.iterrows()],
            dim=0
        )  # T,3,H,W

        gps = torch.tensor(
            rows.loc[:, list(self.gps_cols)].values,
            dtype=torch.float32
        )  # T,2

        label = int(rows.iloc[-1][self.label_col])
        if self.label_is_one_indexed:
            label -= 1

        return images, gps, torch.tensor(label, dtype=torch.long)


def add_relative_gps_and_normalize(df, cfg, fit_stats=None):
    df = df.copy()

    df["rel_lat"] = df[cfg["uav_lat_col"]].astype(float) - df[cfg["bs_lat_col"]].astype(float)
    df["rel_lon"] = df[cfg["uav_lon_col"]].astype(float) - df[cfg["bs_lon_col"]].astype(float)

    cols = ["rel_lat", "rel_lon"]

    if fit_stats is None:
        fit_stats = {
            "min": df[cols].min(),
            "max": df[cols].max(),
        }

    denom = (fit_stats["max"] - fit_stats["min"]).replace(0, 1.0)
    df[cols] = (df[cols] - fit_stats["min"]) / denom

    return df, fit_stats


def make_sequence_indices(df, seq_len=4, episode_col=None):
    seqs = []

    if episode_col is not None and episode_col in df.columns:
        groups = [list(g) for g in df.groupby(episode_col, sort=False).indices.values()]
    else:
        groups = [list(range(len(df)))]

    for g in groups:
        for start in range(0, len(g) - seq_len + 1):
            seqs.append(g[start:start + seq_len])

    return seqs


def split_df_sequential(df, train_ratio=0.7, val_ratio=0.2):
    n = len(df)
    n_train = int(n * train_ratio)
    n_val = int(n * val_ratio)

    train_df = df.iloc[:n_train].reset_index(drop=True)
    val_df = df.iloc[n_train:n_train+n_val].reset_index(drop=True)
    test_df = df.iloc[n_train+n_val:].reset_index(drop=True)

    return train_df, val_df, test_df
```

dataloader build 

```python
df = pd.read_csv(CONFIG["csv_path"])

train_df, val_df, test_df = split_df_sequential(df, train_ratio=0.7, val_ratio=0.2)

train_df, gps_stats = add_relative_gps_and_normalize(train_df, CONFIG, fit_stats=None)
val_df, _ = add_relative_gps_and_normalize(val_df, CONFIG, fit_stats=gps_stats)
test_df, _ = add_relative_gps_and_normalize(test_df, CONFIG, fit_stats=gps_stats)

train_seq = make_sequence_indices(train_df, CONFIG["seq_len"], CONFIG["episode_col"])
val_seq = make_sequence_indices(val_df, CONFIG["seq_len"], CONFIG["episode_col"])
test_seq = make_sequence_indices(test_df, CONFIG["seq_len"], CONFIG["episode_col"])

print("Train seq:", len(train_seq))
print("Val seq:", len(val_seq))
print("Test seq:", len(test_seq))

train_ds = UAVBeamSequenceDataset(
    train_df,
    root=CONFIG["root"],
    image_col=CONFIG["image_col"],
    label_col=CONFIG["label_col"],
    gps_cols=("rel_lat", "rel_lon"),
    seq_indices=train_seq,
    image_size=CONFIG["image_size"],
    label_is_one_indexed=CONFIG["label_is_one_indexed"],
)

val_ds = UAVBeamSequenceDataset(
    val_df,
    root=CONFIG["root"],
    image_col=CONFIG["image_col"],
    label_col=CONFIG["label_col"],
    gps_cols=("rel_lat", "rel_lon"),
    seq_indices=val_seq,
    image_size=CONFIG["image_size"],
    label_is_one_indexed=CONFIG["label_is_one_indexed"],
)

test_ds = UAVBeamSequenceDataset(
    test_df,
    root=CONFIG["root"],
    image_col=CONFIG["image_col"],
    label_col=CONFIG["label_col"],
    gps_cols=("rel_lat", "rel_lon"),
    seq_indices=test_seq,
    image_size=CONFIG["image_size"],
    label_is_one_indexed=CONFIG["label_is_one_indexed"],
)

BATCH_SIZE = 8

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True, drop_last=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

images, gps, labels = next(iter(train_loader))
print("images:", images.shape)   # B,4,3,256,256
print("gps:", gps.shape)         # B,4,2
print("labels:", labels.shape)
print("label min/max:", labels.min().item(), labels.max().item())
```
