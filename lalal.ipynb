```python
from pathlib import Path
from typing import List, Tuple
from functools import lru_cache
import ast
import re
import numpy as np
import pandas as pd
from PIL import Image

import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms


@lru_cache(maxsize=4)
def build_image_index(root_str):
    """
    Root er vitore sob image filename index kore.
    Direct path fail korle filename diye image khuje ber korbe.
    """
    root = Path(root_str)
    image_index = {}

    for ext in ["*.jpg", "*.jpeg", "*.png", "*.JPG", "*.JPEG", "*.PNG"]:
        for p in root.rglob(ext):
            image_index[p.name] = p

    print(f"Image index built from {root}: {len(image_index)} images found")
    return image_index


def auto_fix_root(root):
    """
    Correct dataset root detect kore.
    Correct root holo oi folder jekhane:
        unit1/camera_data
    exists.
    """
    root = Path(root)

    if (root / "unit1" / "camera_data").exists():
        return root

    jpgs = list(root.rglob("*.jpg"))

    if len(jpgs) == 0:
        return root

    first_img = jpgs[0]

    for parent in first_img.parents:
        if (parent / "unit1" / "camera_data").exists():
            return parent

    return root


def resolve_path(root: Path, value: str) -> Path:
    """
    Robust image resolver.

    CSV te image path:
        ./unit1/camera_data/image_xxx.jpg

    Tries:
    1. absolute path
    2. root / cleaned relative path
    3. filename search inside root
    """
    root = auto_fix_root(root)

    value = str(value).strip().replace("\\", "/")
    cleaned = value.replace("./", "")

    p = Path(value)

    if p.is_absolute() and p.exists():
        return p

    candidate = root / cleaned

    if candidate.exists():
        return candidate

    image_index = build_image_index(str(root))
    fname = Path(cleaned).name

    if fname in image_index:
        return image_index[fname]

    raise FileNotFoundError(
        f"\nImage not found!\n"
        f"CSV value: {value}\n"
        f"Root: {root}\n"
        f"Tried: {candidate}\n"
        f"Filename: {fname}\n"
    )


def parse_loc_value(x):
    """
    unit1_loc / unit2_loc parse kore x,y,z ber kore.

    Handles:
    '[x, y, z]'
    '(x, y, z)'
    'x,y,z'
    or string with numbers.
    """
    if isinstance(x, (list, tuple, np.ndarray)):
        vals = list(x)[:3]

        while len(vals) < 3:
            vals.append(0.0)

        return np.array(vals, dtype=np.float32)

    s = str(x).strip()

    try:
        val = ast.literal_eval(s)

        if isinstance(val, (list, tuple)):
            vals = list(val)[:3]

            while len(vals) < 3:
                vals.append(0.0)

            return np.array(vals, dtype=np.float32)

    except Exception:
        pass

    nums = re.findall(r"[-+]?\d*\.\d+|[-+]?\d+", s)
    nums = [float(v) for v in nums]

    if len(nums) >= 3:
        return np.array(nums[:3], dtype=np.float32)

    elif len(nums) == 2:
        return np.array([nums[0], nums[1], 0.0], dtype=np.float32)

    elif len(nums) == 1:
        return np.array([nums[0], 0.0, 0.0], dtype=np.float32)

    return np.array([0.0, 0.0, 0.0], dtype=np.float32)


class UAVBeamSequenceDataset(Dataset):
    """
    Paper-style UAV beam tracking dataset.

    Output:
        images: [T, 3, 256, 256]
        gps:    [T, 2]
        label:  scalar

    T = 4:
        t-3, t-2, t-1, t
    """

    def __init__(
        self,
        df: pd.DataFrame,
        root: str,
        image_col: str,
        label_col: str,
        gps_cols: Tuple[str, str],
        seq_indices: List[List[int]],
        image_size: int = 256,
        label_is_one_indexed: bool = True,
    ):
        self.df = df.reset_index(drop=True)
        self.root = auto_fix_root(Path(root))
        self.image_col = image_col
        self.label_col = label_col
        self.gps_cols = gps_cols
        self.seq_indices = seq_indices
        self.label_is_one_indexed = label_is_one_indexed

        self.transform = transforms.Compose([
            transforms.Resize((image_size, image_size)),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225],
            ),
        ])

    def __len__(self):
        return len(self.seq_indices)

    def _load_image(self, row):
        img_path = resolve_path(self.root, row[self.image_col])
        img = Image.open(img_path).convert("RGB")
        return self.transform(img)

    def __getitem__(self, idx):
        ids = self.seq_indices[idx]
        rows = self.df.iloc[ids]

        images = torch.stack(
            [self._load_image(row) for _, row in rows.iterrows()],
            dim=0
        )  # [T, 3, H, W]

        gps = torch.tensor(
            rows.loc[:, list(self.gps_cols)].values,
            dtype=torch.float32
        )  # [T, 2]

        label = int(rows.iloc[-1][self.label_col])

        # CrossEntropyLoss needs labels 0..63
        # DeepSense beam index normally 1..64
        if self.label_is_one_indexed:
            label -= 1

        return images, gps, torch.tensor(label, dtype=torch.long)


def add_relative_gps_and_normalize(df, cfg, fit_stats=None):
    """
    CSV columns:
        unit1_loc = BS location
        unit2_loc = UAV location

    rel = unit2_loc - unit1_loc

    Model GPS branch expects B x 2,
    so we use relative x,y.
    """
    df = df.copy()

    bs_col = cfg["bs_loc_col"]
    uav_col = cfg["uav_loc_col"]

    bs_xyz = np.stack(df[bs_col].apply(parse_loc_value).values)
    uav_xyz = np.stack(df[uav_col].apply(parse_loc_value).values)

    rel_xyz = uav_xyz - bs_xyz

    df["rel_lat"] = rel_xyz[:, 0]
    df["rel_lon"] = rel_xyz[:, 1]

    cols = ["rel_lat", "rel_lon"]

    if fit_stats is None:
        fit_stats = {
            "min": df[cols].min(),
            "max": df[cols].max(),
        }

    denom = (fit_stats["max"] - fit_stats["min"]).replace(0, 1.0)

    df[cols] = (df[cols] - fit_stats["min"]) / denom
    df[cols] = df[cols].fillna(0.0)

    return df, fit_stats


def make_sequence_indices(df, seq_len=4, episode_col=None):
    """
    Consecutive sequence banay.
    seq_index thakle separate session/episode mix korbe na.
    """
    seqs = []

    if episode_col is not None and episode_col in df.columns:
        groups = [
            list(g)
            for g in df.groupby(episode_col, sort=False).indices.values()
        ]
    else:
        groups = [list(range(len(df)))]

    for g in groups:
        for start in range(0, len(g) - seq_len + 1):
            seqs.append(g[start:start + seq_len])

    return seqs


def split_df_sequential(df, train_ratio=0.7, val_ratio=0.2):
    """
    Strict sequential split by seq_index.

    Same seq_index/session will not be split across train/val/test.

    Split:
        first 70% seq_index -> train
        next 20% seq_index  -> val
        last 10% seq_index  -> test
    """

    if "seq_index" not in df.columns:
        print("WARNING: seq_index not found. Falling back to row-level sequential split.")

        n = len(df)
        n_train = int(n * train_ratio)
        n_val = int(n * val_ratio)

        train_df = df.iloc[:n_train].reset_index(drop=True)
        val_df = df.iloc[n_train:n_train + n_val].reset_index(drop=True)
        test_df = df.iloc[n_train + n_val:].reset_index(drop=True)

        return train_df, val_df, test_df

    seq_values = df["seq_index"].drop_duplicates().tolist()
    n_seq = len(seq_values)

    n_train_seq = int(n_seq * train_ratio)
    n_val_seq = int(n_seq * val_ratio)

    train_seq_values = seq_values[:n_train_seq]
    val_seq_values = seq_values[n_train_seq:n_train_seq + n_val_seq]
    test_seq_values = seq_values[n_train_seq + n_val_seq:]

    train_df = df[df["seq_index"].isin(train_seq_values)].reset_index(drop=True)
    val_df = df[df["seq_index"].isin(val_seq_values)].reset_index(drop=True)
    test_df = df[df["seq_index"].isin(test_seq_values)].reset_index(drop=True)

    print("Strict seq_index-level split:")
    print("Total seq_index:", n_seq)
    print("Train seq_index:", len(train_seq_values))
    print("Val seq_index:", len(val_seq_values))
    print("Test seq_index:", len(test_seq_values))

    print("\nRow counts:")
    print("Train rows:", len(train_df))
    print("Val rows:", len(val_df))
    print("Test rows:", len(test_df))

    # Safety check: no overlap
    train_set = set(train_df["seq_index"].unique())
    val_set = set(val_df["seq_index"].unique())
    test_set = set(test_df["seq_index"].unique())

    assert train_set.isdisjoint(val_set), "Train/Val seq_index overlap!"
    assert train_set.isdisjoint(test_set), "Train/Test seq_index overlap!"
    assert val_set.isdisjoint(test_set), "Val/Test seq_index overlap!"

    print("\nNo seq_index overlap between train/val/test.")

    return train_df, val_df, test_df


print("Cell loaded successfully.")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()

        hidden = max(channels // reduction, 1)

        self.pool = nn.AdaptiveAvgPool2d(1)

        self.fc = nn.Sequential(
            nn.Linear(channels, hidden),
            nn.ReLU(inplace=True),
            nn.Linear(hidden, channels),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _, _ = x.shape

        w = self.pool(x).view(b, c)
        w = self.fc(w).view(b, c, 1, 1)

        return x * w


class BasicBlockSE(nn.Module):
    def __init__(self, in_planes, planes, stride=1):
        super().__init__()

        self.conv1 = nn.Conv2d(
            in_planes,
            planes,
            kernel_size=3,
            stride=stride,
            padding=1,
            bias=False
        )
        self.bn1 = nn.BatchNorm2d(planes)

        self.conv2 = nn.Conv2d(
            planes,
            planes,
            kernel_size=3,
            stride=1,
            padding=1,
            bias=False
        )
        self.bn2 = nn.BatchNorm2d(planes)

        self.se = SEBlock(planes)

        if stride != 1 or in_planes != planes:
            self.shortcut = nn.Sequential(
                nn.Conv2d(
                    in_planes,
                    planes,
                    kernel_size=1,
                    stride=stride,
                    bias=False
                ),
                nn.BatchNorm2d(planes)
            )
        else:
            self.shortcut = nn.Identity()

    def forward(self, x):
        identity = self.shortcut(x)

        out = self.conv1(x)
        out = self.bn1(out)
        out = F.relu(out, inplace=True)

        out = self.conv2(out)
        out = self.bn2(out)

        out = self.se(out)

        out = out + identity
        out = F.relu(out, inplace=True)

        return out


class CrossAttentionBlock(nn.Module):
    """
    CAformer-like bidirectional cross-attention.

    Image branch:
        Query = image feature
        Key/Value = GPS feature

    GPS branch:
        Query = GPS feature
        Key/Value = image feature
    """

    def __init__(self, dim, heads=4, mlp_ratio=2.0, dropout=0.1):
        super().__init__()

        self.img_norm1 = nn.LayerNorm(dim)
        self.gps_norm1 = nn.LayerNorm(dim)

        self.img_attn = nn.MultiheadAttention(
            embed_dim=dim,
            num_heads=heads,
            dropout=dropout,
            batch_first=True
        )

        self.gps_attn = nn.MultiheadAttention(
            embed_dim=dim,
            num_heads=heads,
            dropout=dropout,
            batch_first=True
        )

        self.img_norm2 = nn.LayerNorm(dim)
        self.gps_norm2 = nn.LayerNorm(dim)

        hidden = int(dim * mlp_ratio)

        self.img_mlp = nn.Sequential(
            nn.Linear(dim, hidden),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden, dim)
        )

        self.gps_mlp = nn.Sequential(
            nn.Linear(dim, hidden),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden, dim)
        )

    def forward(self, img_feat, gps_feat):
        """
        img_feat: [B, C, H, W]
        gps_feat: [B, C]
        """

        b, c, h, w = img_feat.shape

        img_token = F.adaptive_avg_pool2d(img_feat, 1).flatten(1).unsqueeze(1)
        gps_token = gps_feat.unsqueeze(1)

        # Image query attends GPS key/value
        img_q = self.img_norm1(img_token)
        gps_kv = self.gps_norm1(gps_token)

        img_delta, _ = self.img_attn(img_q, gps_kv, gps_kv)
        img_token = img_token + img_delta
        img_token = img_token + self.img_mlp(self.img_norm2(img_token))

        # GPS query attends image key/value
        gps_q = self.gps_norm1(gps_token)
        img_kv = self.img_norm1(img_token)

        gps_delta, _ = self.gps_attn(gps_q, img_kv, img_kv)
        gps_token = gps_token + gps_delta
        gps_token = gps_token + self.gps_mlp(self.gps_norm2(gps_token))

        # Recalibrate image feature map using image token
        img_gate = torch.sigmoid(img_token.squeeze(1)).view(b, c, 1, 1)
        img_feat = img_feat + img_feat * img_gate

        gps_feat = gps_token.squeeze(1)

        return img_feat, gps_feat


print("Cell 11 loaded: SEBlock, BasicBlockSE, CrossAttentionBlock")
```

```python
class ResNetSECAformerEncoder(nn.Module):
    def __init__(self, gps_in=2, dropout=0.1):
        super().__init__()

        self.stem = nn.Sequential(
            nn.Conv2d(
                3,
                64,
                kernel_size=7,
                stride=2,
                padding=3,
                bias=False
            ),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(
                kernel_size=3,
                stride=2,
                padding=1
            )
        )

        self.in_planes = 64

        # ResNet34 block pattern: 3, 4, 6, 3
        self.layer1 = self._make_layer(64, blocks=3, stride=1)
        self.layer2 = self._make_layer(128, blocks=4, stride=2)
        self.layer3 = self._make_layer(256, blocks=6, stride=2)
        self.layer4 = self._make_layer(512, blocks=3, stride=2)

        # GPS / location branch: 2 -> 64 -> 128 -> 256 -> 512
        self.gps_fc1 = nn.Linear(gps_in, 64)
        self.gps_fc2 = nn.Linear(64, 128)
        self.gps_fc3 = nn.Linear(128, 256)
        self.gps_fc4 = nn.Linear(256, 512)

        # CAformer blocks
        self.ca1 = CrossAttentionBlock(64, heads=4, dropout=dropout)
        self.ca2 = CrossAttentionBlock(128, heads=4, dropout=dropout)
        self.ca3 = CrossAttentionBlock(256, heads=4, dropout=dropout)
        self.ca4 = CrossAttentionBlock(512, heads=8, dropout=dropout)

        self.pool = nn.AdaptiveAvgPool2d(1)

    def _make_layer(self, planes, blocks, stride):
        layers = []

        layers.append(
            BasicBlockSE(
                in_planes=self.in_planes,
                planes=planes,
                stride=stride
            )
        )

        self.in_planes = planes

        for _ in range(1, blocks):
            layers.append(
                BasicBlockSE(
                    in_planes=self.in_planes,
                    planes=planes,
                    stride=1
                )
            )

        return nn.Sequential(*layers)

    def forward_one_step(self, image, gps):
        """
        image: [B, 3, 256, 256]
        gps:   [B, 2]
        """

        x = self.stem(image)

        # Stage 1
        x = self.layer1(x)
        g = F.relu(self.gps_fc1(gps), inplace=True)
        x, g = self.ca1(x, g)

        # Stage 2
        x = self.layer2(x)
        g = F.relu(self.gps_fc2(g), inplace=True)
        x, g = self.ca2(x, g)

        # Stage 3
        x = self.layer3(x)
        g = F.relu(self.gps_fc3(g), inplace=True)
        x, g = self.ca3(x, g)

        # Stage 4
        x = self.layer4(x)
        g = F.relu(self.gps_fc4(g), inplace=True)
        x, g = self.ca4(x, g)

        img_vec = self.pool(x).flatten(1)  # [B, 512]
        gps_vec = g                        # [B, 512]

        feat = torch.cat([img_vec, gps_vec], dim=1)  # [B, 1024]

        return feat

    def forward(self, images, gps):
        """
        images: [B, T, 3, 256, 256]
        gps:    [B, T, 2]
        """

        b, t, c, h, w = images.shape

        feats = []

        for i in range(t):
            feat_i = self.forward_one_step(
                images[:, i],
                gps[:, i]
            )
            feats.append(feat_i)

        seq = torch.stack(feats, dim=1)  # [B, T, 1024]

        return seq


class UAVBeamTracker(nn.Module):
    def __init__(
        self,
        num_beams=64,
        lstm_hidden=512,
        lstm_layers=4,
        dropout=0.1,
        mlp_hidden=512
    ):
        super().__init__()

        self.encoder = ResNetSECAformerEncoder(
            gps_in=2,
            dropout=dropout
        )

        self.lstm = nn.LSTM(
            input_size=1024,
            hidden_size=lstm_hidden,
            num_layers=lstm_layers,
            batch_first=True,
            dropout=dropout if lstm_layers > 1 else 0.0
        )

        self.classifier = nn.Sequential(
            nn.Linear(lstm_hidden, mlp_hidden),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout),

            nn.Linear(mlp_hidden, mlp_hidden // 2),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout),

            nn.Linear(mlp_hidden // 2, num_beams)
        )

    def forward(self, images, gps):
        seq = self.encoder(images, gps)   # [B, 4, 1024]

        out, _ = self.lstm(seq)

        ot = out[:, -1, :]                # [B, 512]

        logits = self.classifier(ot)      # [B, 64]

        return logits


print("Cell 12 loaded: Full UAVBeamTracker model")
```

```python
@torch.no_grad()
def topk_accuracy(logits, labels, ks=(1, 2, 3)):
    max_k = max(ks)

    _, pred = logits.topk(max_k, dim=1)  # [B, max_k]
    pred = pred.t()                      # [max_k, B]

    correct = pred.eq(labels.view(1, -1).expand_as(pred))

    out = {}

    for k in ks:
        out[f"top{k}"] = (
            correct[:k]
            .reshape(-1)
            .float()
            .sum()
            .item()
            / labels.numel()
        )

    return out


@torch.no_grad()
def dba_score(logits, labels, ks=(1, 2, 3), delta=5):
    max_k = max(ks)

    _, pred = logits.topk(max_k, dim=1)  # [B, max_k]
    labels = labels.view(-1, 1)

    out = {}

    for k in ks:
        pk = pred[:, :k]

        dist = torch.abs(pk - labels)
        min_dist, _ = dist.min(dim=1)

        score = 1.0 - torch.clamp(
            min_dist.float() / delta,
            max=1.0
        )

        out[f"dba{k}"] = score.mean().item()

    return out


class AverageMeter:
    def __init__(self):
        self.reset()

    def reset(self):
        self.sum = 0.0
        self.count = 0
        self.avg = 0.0

    def update(self, val, n=1):
        self.sum += float(val) * n
        self.count += n
        self.avg = self.sum / max(self.count, 1)


print("Cell 13 loaded: metrics ready")
```

```python
from tqdm.auto import tqdm
import torch.optim as optim
from torch.optim.lr_scheduler import MultiStepLR


def run_epoch(model, loader, criterion, optimizer=None):
    is_train = optimizer is not None

    if is_train:
        model.train()
    else:
        model.eval()

    loss_meter = AverageMeter()

    top1_meter = AverageMeter()
    top2_meter = AverageMeter()
    top3_meter = AverageMeter()

    dba1_meter = AverageMeter()
    dba2_meter = AverageMeter()
    dba3_meter = AverageMeter()

    pbar = tqdm(loader, leave=False)

    for images, gps, labels in pbar:
        images = images.to(device, non_blocking=True)
        gps = gps.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        with torch.set_grad_enabled(is_train):
            logits = model(images, gps)

            loss = criterion(logits, labels)

            if is_train:
                optimizer.zero_grad(set_to_none=True)
                loss.backward()

                torch.nn.utils.clip_grad_norm_(
                    model.parameters(),
                    max_norm=5.0
                )

                optimizer.step()

        bsz = labels.size(0)

        acc = topk_accuracy(
            logits,
            labels,
            ks=(1, 2, 3)
        )

        dba = dba_score(
            logits,
            labels,
            ks=(1, 2, 3),
            delta=5
        )

        loss_meter.update(loss.item(), bsz)

        top1_meter.update(acc["top1"], bsz)
        top2_meter.update(acc["top2"], bsz)
        top3_meter.update(acc["top3"], bsz)

        dba1_meter.update(dba["dba1"], bsz)
        dba2_meter.update(dba["dba2"], bsz)
        dba3_meter.update(dba["dba3"], bsz)

        pbar.set_description(
            f"loss={loss_meter.avg:.4f} top1={top1_meter.avg:.4f} top2={top2_meter.avg:.4f}"
        )

    return {
        "loss": loss_meter.avg,

        "top1": top1_meter.avg,
        "top2": top2_meter.avg,
        "top3": top3_meter.avg,

        "dba1": dba1_meter.avg,
        "dba2": dba2_meter.avg,
        "dba3": dba3_meter.avg
    }


print("Cell 15 loaded: run_epoch ready")
```

```python

# Cell 16 — Improved Training


import json
import time
from pathlib import Path
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim.lr_scheduler import MultiStepLR

# -----------------------------
# Training config
# -----------------------------
EPOCHS = 50   # চাইলে 100 করো full run এর জন্য

SAVE_DIR = Path("/content/drive/MyDrive/uav_beam_tracking_outputs_seq_split_improved")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

print("Saving to:", SAVE_DIR)


# -----------------------------
# Class-weighted loss
# -----------------------------
train_labels_raw = train_df[CONFIG["label_col"]].astype(int).values

# Convert 1..64 -> 0..63 if needed
if CONFIG["label_is_one_indexed"]:
    train_labels = train_labels_raw - 1
else:
    train_labels = train_labels_raw

num_classes = CONFIG["codebook_size"]

counts = np.bincount(train_labels, minlength=num_classes)
counts = np.maximum(counts, 1)  # avoid division by zero

class_weights = counts.sum() / (num_classes * counts)
class_weights = torch.tensor(class_weights, dtype=torch.float32).to(device)

print("Class counts min/max:", counts.min(), counts.max())
print("Class weights min/max:", class_weights.min().item(), class_weights.max().item())

criterion = nn.CrossEntropyLoss(weight=class_weights)


# -----------------------------
# Optimizer + Scheduler
# -----------------------------
optimizer = optim.AdamW(
    model.parameters(),
    lr=3e-4,
    weight_decay=1e-4
)

scheduler = MultiStepLR(
    optimizer,
    milestones=[20, 35, 45],
    gamma=0.3
)


# -----------------------------
# Train loop
# -----------------------------
best_top1 = -1.0
history = []

for epoch in range(1, EPOCHS + 1):
    print(f"\nEpoch {epoch}/{EPOCHS}")
    print(f"LR: {optimizer.param_groups[0]['lr']:.2e}")

    start_time = time.time()

    train_metrics = run_epoch(
        model,
        train_loader,
        criterion,
        optimizer=optimizer
    )

    val_metrics = run_epoch(
        model,
        val_loader,
        criterion,
        optimizer=None
    )

    scheduler.step()

    epoch_time = time.time() - start_time

    row = {
        "epoch": epoch,
        "time_sec": epoch_time,
        "train": train_metrics,
        "val": val_metrics
    }

    history.append(row)

    print("Train:", train_metrics)
    print("Val:  ", val_metrics)
    print(f"Epoch time: {epoch_time:.2f} sec")

    print(
        f"Train Top1: {train_metrics['top1'] * 100:.2f}% | "
        f"Val Top1: {val_metrics['top1'] * 100:.2f}%"
    )

    ckpt = {
        "epoch": epoch,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
        "config": CONFIG,
        "history": history,
        "best_top1": best_top1
    }

    torch.save(ckpt, SAVE_DIR / "last.pt")

    if val_metrics["top1"] > best_top1:
        best_top1 = val_metrics["top1"]

        ckpt["best_top1"] = best_top1
        torch.save(ckpt, SAVE_DIR / "best.pt")

        print(f"✅ Saved best.pt | best val top1 = {best_top1 * 100:.2f}%")

# -----------------------------
# Save history
# -----------------------------
with open(SAVE_DIR / "history.json", "w") as f:
    json.dump(history, f, indent=2)

print("\n✅ Training finished.")
print("Best Val Top1:", best_top1 * 100)
print("Saved to:", SAVE_DIR)
```
