```python
from pathlib import Path
from typing import List, Tuple
from functools import lru_cache
import ast
import re
import numpy as np
import pandas as pd
from PIL import Image

import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms


@lru_cache(maxsize=4)
def build_image_index(root_str):
    """
    Root er vitore sob image filename index kore.
    Direct path fail korle filename diye image khuje ber korbe.
    """
    root = Path(root_str)
    image_index = {}

    for ext in ["*.jpg", "*.jpeg", "*.png", "*.JPG", "*.JPEG", "*.PNG"]:
        for p in root.rglob(ext):
            image_index[p.name] = p

    print(f"Image index built from {root}: {len(image_index)} images found")
    return image_index


def auto_fix_root(root):
    """
    Correct dataset root detect kore.
    Correct root holo oi folder jekhane:
        unit1/camera_data
    exists.
    """
    root = Path(root)

    if (root / "unit1" / "camera_data").exists():
        return root

    jpgs = list(root.rglob("*.jpg"))

    if len(jpgs) == 0:
        return root

    first_img = jpgs[0]

    for parent in first_img.parents:
        if (parent / "unit1" / "camera_data").exists():
            return parent

    return root


def resolve_path(root: Path, value: str) -> Path:
    """
    Robust image resolver.

    CSV te image path:
        ./unit1/camera_data/image_xxx.jpg

    Tries:
    1. absolute path
    2. root / cleaned relative path
    3. filename search inside root
    """
    root = auto_fix_root(root)

    value = str(value).strip().replace("\\", "/")
    cleaned = value.replace("./", "")

    p = Path(value)

    if p.is_absolute() and p.exists():
        return p

    candidate = root / cleaned

    if candidate.exists():
        return candidate

    image_index = build_image_index(str(root))
    fname = Path(cleaned).name

    if fname in image_index:
        return image_index[fname]

    raise FileNotFoundError(
        f"\nImage not found!\n"
        f"CSV value: {value}\n"
        f"Root: {root}\n"
        f"Tried: {candidate}\n"
        f"Filename: {fname}\n"
    )


def parse_loc_value(x):
    """
    unit1_loc / unit2_loc parse kore x,y,z ber kore.

    Handles:
    '[x, y, z]'
    '(x, y, z)'
    'x,y,z'
    or string with numbers.
    """
    if isinstance(x, (list, tuple, np.ndarray)):
        vals = list(x)[:3]

        while len(vals) < 3:
            vals.append(0.0)

        return np.array(vals, dtype=np.float32)

    s = str(x).strip()

    try:
        val = ast.literal_eval(s)

        if isinstance(val, (list, tuple)):
            vals = list(val)[:3]

            while len(vals) < 3:
                vals.append(0.0)

            return np.array(vals, dtype=np.float32)

    except Exception:
        pass

    nums = re.findall(r"[-+]?\d*\.\d+|[-+]?\d+", s)
    nums = [float(v) for v in nums]

    if len(nums) >= 3:
        return np.array(nums[:3], dtype=np.float32)

    elif len(nums) == 2:
        return np.array([nums[0], nums[1], 0.0], dtype=np.float32)

    elif len(nums) == 1:
        return np.array([nums[0], 0.0, 0.0], dtype=np.float32)

    return np.array([0.0, 0.0, 0.0], dtype=np.float32)


class UAVBeamSequenceDataset(Dataset):
    """
    Paper-style UAV beam tracking dataset.

    Output:
        images: [T, 3, 256, 256]
        gps:    [T, 2]
        label:  scalar

    T = 4:
        t-3, t-2, t-1, t
    """

    def __init__(
        self,
        df: pd.DataFrame,
        root: str,
        image_col: str,
        label_col: str,
        gps_cols: Tuple[str, str],
        seq_indices: List[List[int]],
        image_size: int = 256,
        label_is_one_indexed: bool = True,
    ):
        self.df = df.reset_index(drop=True)
        self.root = auto_fix_root(Path(root))
        self.image_col = image_col
        self.label_col = label_col
        self.gps_cols = gps_cols
        self.seq_indices = seq_indices
        self.label_is_one_indexed = label_is_one_indexed

        self.transform = transforms.Compose([
            transforms.Resize((image_size, image_size)),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225],
            ),
        ])

    def __len__(self):
        return len(self.seq_indices)

    def _load_image(self, row):
        img_path = resolve_path(self.root, row[self.image_col])
        img = Image.open(img_path).convert("RGB")
        return self.transform(img)

    def __getitem__(self, idx):
        ids = self.seq_indices[idx]
        rows = self.df.iloc[ids]

        images = torch.stack(
            [self._load_image(row) for _, row in rows.iterrows()],
            dim=0
        )  # [T, 3, H, W]

        gps = torch.tensor(
            rows.loc[:, list(self.gps_cols)].values,
            dtype=torch.float32
        )  # [T, 2]

        label = int(rows.iloc[-1][self.label_col])

        # CrossEntropyLoss needs labels 0..63
        # DeepSense beam index normally 1..64
        if self.label_is_one_indexed:
            label -= 1

        return images, gps, torch.tensor(label, dtype=torch.long)


def add_relative_gps_and_normalize(df, cfg, fit_stats=None):
    """
    CSV columns:
        unit1_loc = BS location
        unit2_loc = UAV location

    rel = unit2_loc - unit1_loc

    Model GPS branch expects B x 2,
    so we use relative x,y.
    """
    df = df.copy()

    bs_col = cfg["bs_loc_col"]
    uav_col = cfg["uav_loc_col"]

    bs_xyz = np.stack(df[bs_col].apply(parse_loc_value).values)
    uav_xyz = np.stack(df[uav_col].apply(parse_loc_value).values)

    rel_xyz = uav_xyz - bs_xyz

    df["rel_lat"] = rel_xyz[:, 0]
    df["rel_lon"] = rel_xyz[:, 1]

    cols = ["rel_lat", "rel_lon"]

    if fit_stats is None:
        fit_stats = {
            "min": df[cols].min(),
            "max": df[cols].max(),
        }

    denom = (fit_stats["max"] - fit_stats["min"]).replace(0, 1.0)

    df[cols] = (df[cols] - fit_stats["min"]) / denom
    df[cols] = df[cols].fillna(0.0)

    return df, fit_stats


def make_sequence_indices(df, seq_len=4, episode_col=None):
    """
    Consecutive sequence banay.
    seq_index thakle separate session/episode mix korbe na.
    """
    seqs = []

    if episode_col is not None and episode_col in df.columns:
        groups = [
            list(g)
            for g in df.groupby(episode_col, sort=False).indices.values()
        ]
    else:
        groups = [list(range(len(df)))]

    for g in groups:
        for start in range(0, len(g) - seq_len + 1):
            seqs.append(g[start:start + seq_len])

    return seqs


def split_df_sequential(df, train_ratio=0.7, val_ratio=0.2):
    """
    Strict sequential split by seq_index.

    Same seq_index/session will not be split across train/val/test.

    Split:
        first 70% seq_index -> train
        next 20% seq_index  -> val
        last 10% seq_index  -> test
    """

    if "seq_index" not in df.columns:
        print("WARNING: seq_index not found. Falling back to row-level sequential split.")

        n = len(df)
        n_train = int(n * train_ratio)
        n_val = int(n * val_ratio)

        train_df = df.iloc[:n_train].reset_index(drop=True)
        val_df = df.iloc[n_train:n_train + n_val].reset_index(drop=True)
        test_df = df.iloc[n_train + n_val:].reset_index(drop=True)

        return train_df, val_df, test_df

    seq_values = df["seq_index"].drop_duplicates().tolist()
    n_seq = len(seq_values)

    n_train_seq = int(n_seq * train_ratio)
    n_val_seq = int(n_seq * val_ratio)

    train_seq_values = seq_values[:n_train_seq]
    val_seq_values = seq_values[n_train_seq:n_train_seq + n_val_seq]
    test_seq_values = seq_values[n_train_seq + n_val_seq:]

    train_df = df[df["seq_index"].isin(train_seq_values)].reset_index(drop=True)
    val_df = df[df["seq_index"].isin(val_seq_values)].reset_index(drop=True)
    test_df = df[df["seq_index"].isin(test_seq_values)].reset_index(drop=True)

    print("Strict seq_index-level split:")
    print("Total seq_index:", n_seq)
    print("Train seq_index:", len(train_seq_values))
    print("Val seq_index:", len(val_seq_values))
    print("Test seq_index:", len(test_seq_values))

    print("\nRow counts:")
    print("Train rows:", len(train_df))
    print("Val rows:", len(val_df))
    print("Test rows:", len(test_df))

    # Safety check: no overlap
    train_set = set(train_df["seq_index"].unique())
    val_set = set(val_df["seq_index"].unique())
    test_set = set(test_df["seq_index"].unique())

    assert train_set.isdisjoint(val_set), "Train/Val seq_index overlap!"
    assert train_set.isdisjoint(test_set), "Train/Test seq_index overlap!"
    assert val_set.isdisjoint(test_set), "Val/Test seq_index overlap!"

    print("\nNo seq_index overlap between train/val/test.")

    return train_df, val_df, test_df


print("Cell loaded successfully.")
```

```python

CONFIG = {
    "root": str(root),
    "csv_path": str(CSV_PATH),

    "image_col": "unit1_rgb",
    "label_col": "unit1_beam_index",
    "episode_col": "seq_index",

    "bs_loc_col": "unit1_loc",
    "uav_loc_col": "unit2_loc",

    "image_size": 256,
    "seq_len": 4,
    "codebook_size": 64,

    # Auto-detect below
    "label_is_one_indexed": True,
}

# Auto-fix root so that root/unit1/camera_data exists
CONFIG["root"] = str(auto_fix_root(CONFIG["root"]))

print("CONFIG root:", CONFIG["root"])
print("CSV path:", CONFIG["csv_path"])



df = pd.read_csv(CONFIG["csv_path"])

print("\nData shape:", df.shape)
print("Columns:", list(df.columns))

required_cols = [
    CONFIG["image_col"],
    CONFIG["label_col"],
    CONFIG["episode_col"],
    CONFIG["bs_loc_col"],
    CONFIG["uav_loc_col"],
]

for col in required_cols:
    assert col in df.columns, f"Missing column: {col}"

print("\nRequired columns found.")



label_min = int(df[CONFIG["label_col"]].min())
label_max = int(df[CONFIG["label_col"]].max())

print("\nLabel min/max BEFORE conversion:")
print(label_min, label_max)

if label_min == 0:
    CONFIG["label_is_one_indexed"] = False
elif label_min == 1:
    CONFIG["label_is_one_indexed"] = True
else:
    print("WARNING: label min is neither 0 nor 1. Check label column manually.")

print("label_is_one_indexed:", CONFIG["label_is_one_indexed"])



sample_img_value = df.loc[0, CONFIG["image_col"]]
sample_img_path = resolve_path(Path(CONFIG["root"]), sample_img_value)

print("\nSample CSV image:", sample_img_value)
print("Resolved image path:", sample_img_path)
print("Image exists:", sample_img_path.exists())



train_df, val_df, test_df = split_df_sequential(
    df,
    train_ratio=0.7,
    val_ratio=0.2
)

print("\nSplit dataframe shapes:")
print("train_df:", train_df.shape)
print("val_df:", val_df.shape)
print("test_df:", test_df.shape)



train_df, gps_stats = add_relative_gps_and_normalize(
    train_df,
    CONFIG,
    fit_stats=None
)

val_df, _ = add_relative_gps_and_normalize(
    val_df,
    CONFIG,
    fit_stats=gps_stats
)

test_df, _ = add_relative_gps_and_normalize(
    test_df,
    CONFIG,
    fit_stats=gps_stats
)

print("\nRelative GPS sample:")
display(train_df[["rel_lat", "rel_lon"]].head())



train_seq = make_sequence_indices(
    train_df,
    seq_len=CONFIG["seq_len"],
    episode_col=CONFIG["episode_col"]
)

val_seq = make_sequence_indices(
    val_df,
    seq_len=CONFIG["seq_len"],
    episode_col=CONFIG["episode_col"]
)

test_seq = make_sequence_indices(
    test_df,
    seq_len=CONFIG["seq_len"],
    episode_col=CONFIG["episode_col"]
)

print("\nSequence counts:")
print("Train seq:", len(train_seq))
print("Val seq:", len(val_seq))
print("Test seq:", len(test_seq))

assert len(train_seq) > 0, "No train sequences generated!"
assert len(val_seq) > 0, "No val sequences generated!"
assert len(test_seq) > 0, "No test sequences generated!"

train_ds = UAVBeamSequenceDataset(
    train_df,
    root=CONFIG["root"],
    image_col=CONFIG["image_col"],
    label_col=CONFIG["label_col"],
    gps_cols=("rel_lat", "rel_lon"),
    seq_indices=train_seq,
    image_size=CONFIG["image_size"],
    label_is_one_indexed=CONFIG["label_is_one_indexed"],
)

val_ds = UAVBeamSequenceDataset(
    val_df,
    root=CONFIG["root"],
    image_col=CONFIG["image_col"],
    label_col=CONFIG["label_col"],
    gps_cols=("rel_lat", "rel_lon"),
    seq_indices=val_seq,
    image_size=CONFIG["image_size"],
    label_is_one_indexed=CONFIG["label_is_one_indexed"],
)

test_ds = UAVBeamSequenceDataset(
    test_df,
    root=CONFIG["root"],
    image_col=CONFIG["image_col"],
    label_col=CONFIG["label_col"],
    gps_cols=("rel_lat", "rel_lon"),
    seq_indices=test_seq,
    image_size=CONFIG["image_size"],
    label_is_one_indexed=CONFIG["label_is_one_indexed"],
)


BATCH_SIZE = 8

# Debugging stable rakhar jonno num_workers=0.
# Later speed er jonno num_workers=2 dite paren.
train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=False,
    drop_last=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=False
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=False
)



images, gps, labels = next(iter(train_loader))

print("\nFinal batch check:")
print("images:", images.shape)   # [8, 4, 3, 256, 256]
print("gps:", gps.shape)         # [8, 4, 2]
print("labels:", labels.shape)   # [8]
print("label min/max batch:", labels.min().item(), labels.max().item())
print("gps min/max batch:", gps.min().item(), gps.max().item())

assert images.shape == (
    BATCH_SIZE,
    CONFIG["seq_len"],
    3,
    CONFIG["image_size"],
    CONFIG["image_size"]
)

assert gps.shape == (
    BATCH_SIZE,
    CONFIG["seq_len"],
    2
)

assert labels.ndim == 1
assert labels.min().item() >= 0, "Label became negative. Check label_is_one_indexed."
assert labels.max().item() < CONFIG["codebook_size"], "Label exceeds codebook size."

print("\nDataLoader ready with strict seq_index-level split.")
```

```python
from pathlib import Path

old_root = Path(CONFIG["root"])

jpgs = list(old_root.rglob("*.jpg"))
print("Total jpg:", len(jpgs))
print("First jpg:", jpgs[0])

def find_dataset_root_from_image(img_path):
    """
    CSV image path starts like:
        ./unit1/camera_data/image_xxx.jpg
    
    So correct root is the folder that contains:
        unit1/camera_data/
    """
    img_path = Path(img_path)

    for parent in img_path.parents:
        if (parent / "unit1").exists():
            if (parent / "unit1" / "camera_data").exists():
                return parent

    return None

correct_root = find_dataset_root_from_image(jpgs[0])

print("Old CONFIG root:", CONFIG["root"])
print("Detected correct root:", correct_root)

CONFIG["root"] = str(correct_root)

print("Updated CONFIG root:", CONFIG["root"])
```
